```python
# IMPORT LIBRARY
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
# MEMBACA CITRA KTM
img = cv2.imread('KTM SAYA.jpg')

# Konversi BGR ke RGB untuk ditampilkan dengan Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Menampilkan citra awal
plt.figure(figsize=(5, 3))
plt.imshow(img_rgb)
plt.title("Citra KTM")
plt.axis('off')
plt.show()

print("Ukuran citra:", img_rgb.shape)
```

```python
# TUGAS PRAKTIKUM PENERAPAN 1
# Menampilkan kombinasi kanal Merah-Biru dan Hijau-Biru

# Jawaban 1a: Membuat salinan citra
merah_biru = img_rgb.copy()
hijau_biru = img_rgb.copy()

# Jawaban 1b: Kombinasi Merah-Biru
# Kanal Hijau (G) dinolkan
merah_biru[:, :, 1] = 0

# Jawaban 1c: Kombinasi Hijau-Biru
# Kanal Merah (R) dinolkan
hijau_biru[:, :, 0] = 0

# Jawaban 1d: Menampilkan kedua hasil berdampingan
plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
plt.imshow(merah_biru)
plt.title("Kanal Merah - Biru")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(hijau_biru)
plt.title("Kanal Hijau - Biru")
plt.axis('off')

plt.show()
```

```python
# TUGAS PRAKTIKUM PENERAPAN 2
# Menampilkan tiga bentuk pencerminan citra

# Jawaban 2a: Pencerminan vertikal (atas-bawah)
flip_vertikal = cv2.flip(img_rgb, 0)

# Jawaban 2b: Pencerminan horizontal (kiri-kanan)
flip_horizontal = cv2.flip(img_rgb, 1)

# Jawaban 2c: Pencerminan vertikal dan horizontal
flip_keduanya = cv2.flip(img_rgb, -1)

# Jawaban 2d: Menampilkan ketiga hasil dalam satu figure
plt.figure(figsize=(12, 3))

plt.subplot(1, 3, 1)
plt.imshow(flip_vertikal)
plt.title("Vertikal")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(flip_horizontal)
plt.title("Horizontal")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(flip_keduanya)
plt.title("Vertikal + Horizontal")
plt.axis('off')

plt.show()
```

```python
# TUGAS PRAKTIKUM - ANALISA
# Menganalisis perubahan rata-rata intensitas setelah resize

# Mengubah citra menjadi grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Mengambil ukuran awal
h, w = gray.shape

# Menghitung rata-rata intensitas citra awal
rata_awal = gray.mean()

# Mengecilkan citra menjadi setengah ukuran
gray_kecil = cv2.resize(gray, (w // 2, h // 2))

# Membesarkan kembali ke ukuran semula
gray_besar = cv2.resize(gray_kecil, (w, h))

# Menghitung rata-rata intensitas setelah resize
rata_akhir = gray_besar.mean()

# Menampilkan data hasil
print("Ukuran awal              :", gray.shape)
print("Ukuran setelah diperkecil:", gray_kecil.shape)
print("Ukuran setelah diperbesar:", gray_besar.shape)

print("\nRata-rata intensitas awal :", rata_awal)
print("Rata-rata intensitas akhir:", rata_akhir)
print("Selisih intensitas        :", abs(rata_awal - rata_akhir))

# Menampilkan citra awal dan hasil resize
plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Citra Awal")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gray_besar, cmap='gray')
plt.title("Setelah Resize")
plt.axis('off')

plt.show()
```

```python
# TUGAS PRAKTIKUM - APLIKASI
# STUDI KASUS 1
# Membaca dua citra dengan ukuran dan orientasi berbeda

# Membaca citra
bromo = cv2.imread('BROMO.jpeg')
sawah = cv2.imread('SAWAH.jpeg')

# Menampilkan ukuran citra
print("Ukuran citra Bromo :", bromo.shape)
print("Ukuran citra Sawah :", sawah.shape)
```

```python
# STUDI KASUS 1
# Menampilkan kedua citra asli

# Konversi BGR ke RGB
bromo_rgb = cv2.cvtColor(bromo, cv2.COLOR_BGR2RGB)
sawah_rgb = cv2.cvtColor(sawah, cv2.COLOR_BGR2RGB)

# Menampilkan citra
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(bromo_rgb)
plt.title("Bromo - Portrait")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sawah_rgb)
plt.title("Sawah - Landscape")
plt.axis('off')

plt.show()
```

```python
# Membuat fungsi label identitas adaptif

def tambah_label(img, nama):
    # Membuat salinan citra agar citra asli tidak berubah
    hasil = img.copy()

    # Mengambil tinggi dan lebar citra
    h, w = hasil.shape[:2]

    # Menentukan ukuran secara adaptif berdasarkan ukuran citra
    tinggi_bar = int(h * 0.08)
    font_scale = w / 2000
    thickness = max(1, int(w / 1000))

    # Membuat bar hitam di bagian bawah citra
    cv2.rectangle(
        hasil,
        (0, h - tinggi_bar),
        (w, h),
        (0, 0, 0),
        -1
    )

    # Menambahkan teks identitas
    cv2.putText(
        hasil,
        nama,
        (int(w * 0.03), h - int(tinggi_bar * 0.3)),
        cv2.FONT_HERSHEY_SIMPLEX,
        font_scale,
        (255, 255, 255),
        thickness,
        cv2.LINE_AA
    )

    return hasil
```

```python
# Menerapkan fungsi pada dua citra

bromo_label = tambah_label(bromo, "Dokumentasi Bromo - 2026")
sawah_label = tambah_label(sawah, "Dokumentasi Sawah - 2026")
```

```python
# Menampilkan hasil label adaptif pada dua citra

# Konversi hasil BGR ke RGB untuk Matplotlib
bromo_label_rgb = cv2.cvtColor(bromo_label, cv2.COLOR_BGR2RGB)
sawah_label_rgb = cv2.cvtColor(sawah_label, cv2.COLOR_BGR2RGB)

# Menampilkan hasil
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(bromo_label_rgb)
plt.title("Bromo dengan Label")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sawah_label_rgb)
plt.title("Sawah dengan Label")
plt.axis('off')

plt.show()
```

```python
# TUGAS PRAKTIKUM - APLIKASI
# STUDI KASUS 2
# Membaca citra Bromo

bromo = cv2.imread('BROMO.jpeg')

# Konversi BGR ke RGB
bromo_rgb = cv2.cvtColor(bromo, cv2.COLOR_BGR2RGB)

# Menampilkan citra asli
plt.figure(figsize=(4, 5))
plt.imshow(bromo_rgb)
plt.title("Citra Bromo Asli")
plt.axis('off')
plt.show()

print("Ukuran citra asli:", bromo.shape)
```

```python
# Membuat thumbnail Bromo menjadi persegi 1:1

# Mengambil tinggi dan lebar citra
h, w = bromo.shape[:2]

# Menentukan ukuran sisi persegi
sisi = max(h, w)

# Membuat kanvas putih
thumbnail = np.full((sisi, sisi, 3), 255, dtype=np.uint8)

# Menghitung posisi agar foto berada di tengah
x = (sisi - w) // 2
y = (sisi - h) // 2

# Memasukkan foto Bromo ke tengah kanvas
thumbnail[y:y+h, x:x+w] = bromo
```

```python
# Menambahkan identitas pada thumbnail

font_scale = sisi / 2500
tebal = max(1, sisi // 1500)

cv2.putText(
    thumbnail,
    "Explore Indonesia",
    (int(sisi * 0.03), int(sisi * 0.95)),
    cv2.FONT_HERSHEY_SIMPLEX,
    font_scale,
    (0, 0, 0),
    tebal,
    cv2.LINE_AA
)
```

```python
# Menampilkan hasil thumbnail

thumbnail_rgb = cv2.cvtColor(thumbnail, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(5, 5))
plt.imshow(thumbnail_rgb)
plt.title("Thumbnail Bromo 1:1")
plt.axis('off')
plt.show()

print("Ukuran citra asli     :", bromo.shape)
print("Ukuran hasil thumbnail:", thumbnail.shape)
```
